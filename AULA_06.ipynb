Exercício 2: Correção de Erro de Lógica

```python
def soma_pares(lista_numeros):
  total = 0
  for numero in lista_numeros:
    if numero % 2 != 0:
      total += numero
  return total

print(soma_pares([1, 2, 3, 4, 5, 6]))
```

Para que a função condiga exatamente com o nome soma_pares e some apenas os números pares (2, 4 e 6), a condição do if deveria ser alterada para numero % 2 == 0.

Exercício 3: Refatoração para List Comprehension

```python
def filtra_e_dobra(numeros):
  return [num * 2 for num in numeros if num > 5]

print(filtra_e_dobra([1, 6, 3, 8, 2, 10]))
```

 Isso nos permite filtrar e transformar os elementos da lista em uma única linha de código, tornando a função muito mais concisa e legível.

Exercício 4: Refatoração de Código Repetitivo

```python
def processa_dados(dados, operacao):
  return [operacao(x) for x in dados]

lista_a = [1, 2, 3]
lista_b = [4, 5, 6]
lista_c = [7, 8, 9]

# Usamos lambdas para definir a operação dinamicamente
print(processa_dados(lista_a, lambda x: x * 2))
print(processa_dados(lista_b, lambda x: x + 10))
print(processa_dados(lista_c, lambda x: x - 5))
```

Reduzimos consideravelmente as linhas de código redundantes, facilitando futuras alterações e tornando a base de código muito mais limpa e extensível!

Exercício 5: Refatoração com Cláusulas de Guarda (Guard Clauses)

```python
def processar_pedido(status, quantidade, estoque):
  if status != "aprovado":
    return "Pedido não aprovado."

  if quantidade <= 0:
    return "Quantidade inválida."

  if estoque < quantidade:
    return "Não há estoque suficiente."

  return "Pedido processado com sucesso!"

print(processar_pedido("aprovado", 5, 10))
print(processar_pedido("aprovado", 5, 3))
print(processar_pedido("reprovado", 5, 10))
```

Se o fluxo da função passar por todas essas validações sem ser interrompido, significa que o pedido está perfeitamente correto. Portanto, ao final, basta colocar o caso de sucesso direto: return "Pedido processado com sucesso!" de forma limpa e sem nenhuma indentação adicional.

Exercício 6: Gerenciamento de Recursos com with Statement

```python
def ler_arquivo_seguro(nome_arquivo):
  with open(nome_arquivo, "r") as f:
    return f.read()

with open("teste.txt", "w") as f:
  f.write("Olá, mundo!")

print(ler_arquivo_seguro("teste.txt"))
```

Vamos refatorar a função de leitura para utilizar o bloco with. Dessa forma, garantimos que o arquivo será fechado corretamente, mesmo se ocorrer um erro durante a leitura do conteúdo.

Exercício 7: Simplificação de Lógica Condicional Complexa

```python
def obter_desconto(tipo_cliente, valor_compra):
  regras_desconto = {
      "premium": (1000, 0.20, 0.10),
      "gold": (500, 0.15, 0.05)
  }

  if tipo_cliente not in regras_desconto:
    return 0

  limite, desc_alto, desc_baixo = regras_desconto[tipo_cliente]
  percentual = desc_alto if valor_compra > limite else desc_baixo
  return valor_compra * percentual

print(obter_desconto("premium", 1200))
print(obter_desconto("gold", 600))
print(obter_desconto("normal", 400))
```

Para simplificar essa lógica condicional cheia de if aninhados, podemos mapear as regras de desconto de cada tipo de cliente em uma estrutura de dados mais limpa, como um dicionário, ou nivelar as validações usando cláusulas de guarda.

Exercício 8: Extração de Função (Extract Method)

```python
class ProcessadorDeDados:
  def __init__(self, dados):
    self.dados = dados

  def _validar_dados(self):
    return isinstance(self.dados, list) and len(self.dados) > 0

  def _filtrar_dados(self):
    return [item for item in self.dados if isinstance(item, int) and item > 0]

  def _transformar_dados(self, dados_filtrados):
    return [item * 10 for item in dados_filtrados]

  def _exibir_relatorio(self, dados_transformados):
    print("--- Relatório de Processamento ---")
    print(f"Total de itens processados: {len(dados_transformados)}")
    print(f"Dados transformados: {dados_transformados}")
    print("----------------------------------")

  def processar_dados(self):
    if not self._validar_dados():
      print("Erro: dados inválidos")
      return []

    dados_filtrados = self._filtrar_dados()
    dados_transformados = self._transformar_dados(dados_filtrados)
    self._exibir_relatorio(dados_transformados)

    return dados_transformados

proc = ProcessadorDeDados([1, 2, -3, 4, "a", 5])
proc.processar_dados()
```

O método processar_dados está sobrecarregado: ele valida a entrada, filtra os dados, faz a transformação dos números e ainda imprime o relatório de processamento. Além disso, o código original possui um erro de indentação nas linhas de impressão e no return.
