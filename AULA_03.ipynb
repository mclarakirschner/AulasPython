CRIAÇÃO E MANIPULAÇÃO DE LISTAS []

```python
alunos = []
alunos = ["Ana Julia","Gabriel","Davi","Lucas"]
```

```python
# Contando o tamanho da lista através do len
print(f"Tamanho da lista: {len(alunos)}")
```

```python
# Exemplo: usar a lista toda
print(alunos)

#Exemplo acessar somente 1
print(alunos[1])
```

```python
# Exemplo modificar item existente
alunos[1] = "Gustavo"
```

```python
# Exemplo: Adicionar na lista novo aluno
alunos.append("Gabriel")

print(alunos)
print(len(alunos))
```

```python
# Exemplo: iterar/percorrer listas (olhar um item porvez em sequência)
for aluno in alunos:
  print(aluno)
```

```python
alunos_caixa_alta = []

for aluno in alunos:
  alunos_caixa_alta.append(aluno.upper())

print(alunos_caixa_alta)
```

EXEMPLO - FILTRAR VALORES DE LISTAS

```python
frase = "Eu sei programar em Java, Python, C#, C e Ruby"

# Antes de filtrar remover eventuias pontuações
frase.replace(",","")
tokens = frase.split()

# Filtro
linguagens_encontradas = []
linguagens_mapeadas = ["Java", "Python", "C", "C#", "Ruby", "Go"]


for token in tokens:
  if token not in linguagens_mapeadas:
    linguagens_encontradas.append(token)

print(linguagens_encontradas)
```

EXEMPLO - SINTAXE LIST COMPREHENSION

```python
frase = "Eu sei programar em Java, Python, C#, C e Ruby"

# Antes de filtrar remover eventuias pontuações
frase.replace(",","")
tokens = frase.split()

linguagens_mapeadas = ["Java", "Python", "C", "C#", "Ruby", "Go"]

linguagens_encontradas = [token for token in tokens if token not in linguagens_mapeadas]
print(linguagens_encontradas)
```
