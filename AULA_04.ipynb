Criar e identificar

```python
respostas = {}
respostas = {
    "Curso IA" : "Acontece na terça-feira",
    "Curso Power BI" : "Sem turmas abertas",
    "Curso Metrologia" : "Curso Descontinuado"
}
```

```python
# Como acessar um valor
print(respostas["Curso IA"])
```

```python
# Adicionar nova resposta
respostas["Curso Marketing"] = "Aulas aos sábados"
print(respostas)
```

Iteração (percorrer)

```python
# Isso acessa somente a chave não o valor
for resposta in respostas:
  print(resposta)
```

```python
# Isso acessa a chave e o valor
for chave, valor in respostas.items():
  print(chave)
  print(valor)
```

Exemplo dicionários com listas

```python
# Valores com listas, o for procura o item nas duas listas
sinonimos = {
    "celular" : ["Smartphone","Telemóvel"],
    "cadeira" : ["Assento", "Banco"]
}

# 1. Percorro o dicionario (Lista por lista)
for chave, valor in sinonimos.items():
  # Percorrer item por item da lista atual
  for item in valor:
    print (item)


# 2. Percorrer os elementos
for chave, valor in sinonimos.items():
  if "Banco" in valor:
    print(f"\nValor encontrado na chave {chave} ")
```
