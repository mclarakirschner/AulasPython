```python
import nltk
from nltk.tokenize import word_tokenize # Função que vai criar os tokens (tokenização)
from nltk.corpus import stopwords # Lista pronta de stopWords em portugues
from nltk.stem import RSLPStemmer # Algoritimo que reconhece os stemnings
```

```python
nltk.download('rslp') # algorítimo de stemização
nltk.download('stopwords') # stopwords
nltk.download('punkt_tab') #tokenização
```

2. Função para processamento

```python
def identificar_tokens(mensagem):
  mensagem_padronizada = mensagem.lower()
  tokens = word_tokenize(mensagem_padronizada)
  return tokens
```

```python
def remover_stopwords(tokens):
  stop_words_ptbr = set(stopwords.words('portuguese'))
  return [
      token
      for token in tokens
      if (token not in stop_words_ptbr) and token.isalpha()
  ]
```

```python
def criar_stemmings(tokens):
  stemmer = RSLPStemmer()
  return [stemmer.stem(token) for token in tokens]
```

```python
entidades = {
    "maquina": ["máquina", "maquin", "equipamento", "ativos", "maquinario", "maquinarios", "sistema",],
    "manutencao": ["ajuste","reparo","manutenção","consertar", "reparar", "arrumar", "revisar", "inspecionar", "ajustar", "calibrar",]
}

def identificar_entidade(lista_stemmings):
  for token in lista_stemmings:
    for entidade, lista_sinonimos in entidades.items():
      if token in lista_sinonimos:
        return entidade
```

```python
intencoes = {
    "comprar": ["compr", "vend", "adquir"],
    "consultar": ["consult", "ver", "disponi", "disponibili"],
    "agendar": ["agend", "program", "marc"],
    "cancelar": ["cancel", "anul", "abort", "invalid"]
}

def identificar_intencao(lista_stemmings):
  for token in lista_stemmings:
    for intencao, lista_sinonimos in intencoes.items():
      if token in lista_sinonimos:
        return intencao
```

```python
matriz_resposta = {
    "maquina": {
        "comprar": "Mando o PIX na chave XXX",
        "consultar": "Segue o nosso catálogo em anexo",
    },
    "manutencao": {
        "agendar": "Iforme o local e o horário",
        "cancelar": "Agendamento cancelado com sucesso"
      }
}

def responder(entidade, intencao):
  if (entidade in matriz_resposta) and (intencao in matriz_resposta[entidade]):
    return matriz_resposta[entidade][intencao]
  return "Não entendi sua mensagem"
```

3. Função de execução

```python
def processar_mensagem(mensagem):
  # 1. Tokenização
  tokens = identificar_tokens(mensagem)
  print(f"TOKENS CRIADOS: {tokens}")

  # 2. Remover stopwords
  tokens_filtrados = remover_stopwords(tokens)
  print(f"TOKENS FILTRADOS: {tokens_filtrados}")

  # 3. Criar stemmings
  tokens_steammings = criar_stemmings(tokens_filtrados)
  print(f"TOKENS COM STEMMINGS: {tokens_steammings}")

  # 4. Identificar entidade
  entidade = identificar_entidade(tokens_steammings)
  print(f"ENTIDADE: {entidade}")

  # 5. Identificar a intenção
  intencao = identificar_intencao(tokens_steammings)
  print(f"INTENÇÃO: {intencao}")

  # 6. Gerar resposta
  return responder(entidade, intencao)
```

```python
print("Olá, eu sou o chatbot da GPSA - Manuteções de Máquinas!")
mensagem = input("Digite a sua mensagem: ")
resposta = processar_mensagem(mensagem)
print(resposta)
```
