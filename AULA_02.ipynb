Método 1 -> len() -> Conta o número de caracteres da string

```python
nome = "Maria Clara"

tamanho_texto = len("Maria Clara")
print(tamanho_texto)
```

Métodos de Formatação: upper() e lower()

```python
linguagem = "pYTHON"

# PADRONIZAR EM CAIXA ALTA
linguagem = linguagem.upper()
print(linguagem)
```

```python
linguagem = "pYTHON"

# PADRONIZAR EM CAIXA BAIXA
linguagem = linguagem.lower()
print(linguagem)
```

Método 3: strip() -> remove espaços no começo ou no fim de uma string

```python
nome = 'Ana Julia '
print(nome) # <- valor antes da remoção
print(len(nome)) # <- tamanho antes da recepção


nome = nome.strip() # remove os espaços antes ou depois
print(nome) # <- valor depois da remoção
print(len(nome)) # <- tamanho depois da recepção
```

Método S: replace() -> substituir letras ou caracteres

```python
nome = "Ana"

# Passo a passo
nome = nome.lower() # deixa tudo em minusculo
nome = nome.replace("a" , "u") # substitui a letra a pela letra u
print(nome)

# Fazendo de uma forma direta
nome = nome.lower().replace("a" , "u")
print(nome)
```

```python
# Outro exemplo
nome = "Ana Julia"
nome = nome.replace("Ana" , "Maria")
print(nome)
```

```python
entrada = "      Atenção: o produto BANANA esta em PROMOÇÃO!          "

entrada = entrada.strip()
entrada = entrada.upper()
entrada = entrada.replace("BANANA", "MAÇA")

#OU

entrada = entrada.strip().upper().replace("BANANA", "MAÇA")
print(entrada)

# para contar os caracteres
print(len(entrada))
```

Iteração de strings: percorre subpartes

```python
# CASO 1: Percorrer caracter por caracter
frase = "Hoje está um lindo dia!"

for caracter in frase:
  print(caracter.upper())

```

```python
# CASO 2: Percorrer palavra por palavra
frase = "Hoje está um lindo dia!"

frase = frase.split()

for palavra in frase:
  print(palavra)
```
